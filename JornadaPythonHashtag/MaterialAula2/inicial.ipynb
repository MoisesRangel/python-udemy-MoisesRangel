### Case - Cancelamento de Clientes

Você foi contratado por uma empresa com mais de 800 mil clientes para um projeto de Dados. Recentemente a empresa percebeu que da sua base total de clientes, a maioria são clientes inativos, ou seja, que já cancelaram o serviço.

Precisando melhorar seus resultados ela quer conseguir entender os principais motivos desses cancelamentos e quais as ações mais eficientes para reduzir esse número.

Base de dados e arquivos: https://drive.google.com/drive/folders/1uDesZePdkhiraJmiyeZ-w5tfc8XsNYFZ?usp=drive_link

```python
# passo a passo do projeto
# passo 1 : abrir a base de dados (importar a base de dados)

import pandas as pd

tabela = pd.read_csv("C:/Users/moise/OneDrive/Documentos/ProjetoPython/python-udemy-MoisesRangel/JornadaPythonHashtag/MaterialAula2/cancelamentos.csv")

tabela = tabela.drop(columns='CustomerID')
display(tabela)


# passo 2 : visualizar a base de dados
#   entender as informações
#   possiveis problemas/erros na base de dados
# passo 3 : corrigir os problemas da base de dados (tratamento de dados)
# passo 4 : analise inicial (entender quantos clientes cancelaram)
# passo 5 : analise detalhada (causa do cancelamento dos clientes, como cada coluna impacta no cancelamento)
```

```python
# passo 3 : corrigir os problemas da base de dados (tratamento de dados)
# Valores em formatos errados -> OK
# Informações vazias


display(tabela.info())

tabela = tabela.dropna()
# tabela = tabela.drop_duplicates()

display(tabela.info())
```

```python
# passo 4 : analise inicial (entender quantos clientes cancelaram)

# quantidade - contar quantos caras são 0 e quantos sao 1 na coluna cancelou

display(tabela['cancelou'].value_counts())
display(tabela['duracao_contrato'].value_counts())

# porcentagem
display(tabela['cancelou'].value_counts(normalize=True))

#display(tabela['cancelou'].value_counts(normalize=True).map('{:.1%}'.format))
```

```python
# passo 5 : analise detalhada (causa do cancelamento dos clientes, como cada coluna impacta no cancelamento)
# Melhor forma de ver é com gráficos, então apresentar no formato gráfico facilita o entendimento.

import plotly.express as px
# Primeiro voce cria o grafico
for coluna in tabela.columns:
    grafico = px.histogram(tabela, x=coluna, color='cancelou')
# Segundo voce exibe o grafico
    grafico.show()
```

```python
#São analises de dados o que você consegue extrair dos dados, como por exemplo: 
# todo mundo do contrato mensal, cancelou o serviço - analise
# vamos dar desconto no contrato anual, trimestral - solução
# ligações call center a cima de 4, todo mundo cancelou - analise
#   tem algum problema que nao estamos conseguindo resolver - analise
#   se o cliente ligar 3x pro callcenter, alerta vermelho - solucao
# atraso no pagamento acima d e20 dias, cliente cancela - analise
#   se o cliente atrasar 15 dias no pagamento, alerta vermelho - solucao

#cancelamento = 56%

# Filtrar a base de dados
#duracao_contrato - diferente de monthly
display(tabela['cancelou'].value_counts(normalize=True))

condicao = tabela['duracao_contrato'] != 'Monthly'

tabela = tabela[condicao]

#dias_atraso - menores ou iguais a 20

condicao = tabela['dias_atraso'] <= 20
tabela = tabela[condicao]

#ligacoes_callcenter - menores ou iguais a 4

condicao = tabela['ligacoes_callcenter'] <= 4
tabela = tabela[condicao]

display(tabela['cancelou'].value_counts(normalize=True))
```
